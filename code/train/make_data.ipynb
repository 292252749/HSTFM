```python
import os
import shutil

# 源文件夹
source_folder = r"/mnt/sdg-gpt/chen/code/MLFF-GAN-main/data_CIA/"

# 目标文件夹
target_folder_A = r"/mnt/sdg-gpt/chen/code/Deep-Learning-Spatiotemporal-Fusion-Survey-main/data/CIA/Landsat/"
target_folder_B = r"/mnt/sdg-gpt/chen/code/Deep-Learning-Spatiotemporal-Fusion-Survey-main/data/CIA/MODIS/"

# 如果目标文件夹不存在，则创建
for folder in [target_folder_A, target_folder_B]:
    if not os.path.exists(folder):
        os.makedirs(folder)
        print(f"已创建文件夹: {folder}")

# 遍历源文件夹及其所有子文件夹
for root, dirs, files in os.walk(source_folder):
    for filename in files:
        if filename.lower().endswith(".tif"):
            source_file = os.path.join(root, filename)

            # 判断文件名并复制到对应文件夹
            if "HRF_modtran" in filename:
                target_file = os.path.join(target_folder_A, filename)
                if not os.path.exists(target_file):
                    shutil.copy2(source_file, target_file)
                    print(f"已复制到 A 文件夹: {filename}")
                else:
                    print(f"跳过（已存在）: {filename}")

            elif "MOD09GA" in filename:
                target_file = os.path.join(target_folder_B, filename)
                if not os.path.exists(target_file):
                    shutil.copy2(source_file, target_file)
                    print(f"已复制到 B 文件夹: {filename}")
                else:
                    print(f"跳过（已存在）: {filename}")

print("分类复制完成！")
```

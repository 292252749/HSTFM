```python
from shapely.geometry import Point
import geopandas as gpd
from pyproj import Transformer
from mgrs import MGRS

m = MGRS()

ocean_shp= "/data1/process/s23_20251109_20m/shp/ne_10m_ocean.shp"
oceans = gpd.read_file(ocean_shp)
if oceans.crs is None:
    oceans = oceans.set_crs("EPSG:4326", allow_override=True)
    
def is_ocean_major(tile_id, oceans):
    tile_id = tile_id[1:]
    # 先尝试 50km 正中心（更稳）
    mid = tile_id + "5000050000"
    try:
        lat, lon = m.toLatLon(mid)
    except:
        # 再尝试下一级精度
        mid = tile_id + "5555555555"
        lat, lon = m.toLatLon(mid)

    pt_wgs84 = Point(lon, lat)

    if oceans.crs.to_string() != "EPSG:4326":
        tf = Transformer.from_crs("EPSG:4326", oceans.crs, always_xy=True)
        x, y = tf.transform(lon, lat)
        pt = Point(x, y)
    else:
        pt = pt_wgs84

    try:
        idx = oceans.sindex
        cand = oceans.iloc[list(idx.intersection(pt.bounds))]
    except:
        cand = oceans

    for g in cand.geometry:
        if g.contains(pt):
            return True
    return False
```

```python
import concurrent.futures
from tqdm import tqdm
from osgeo import gdal

def get_s2_cloud(zip_file):
    s = gdal.Open(zip_file)
    metadata = s.GetMetadata()
    return float(metadata['CLOUD_COVERAGE_ASSESSMENT'])


def process_one_key(args):
    key, value, oceans = args

    if is_ocean_major(key, oceans):
        return key, None, []

    if len(value) <= 1:
        return key, None, []

    zip_files = [v[0] for v in value]

    # key 内部顺序处理
    clouds = [get_s2_cloud(z) for z in zip_files]

    items = list(zip(value, clouds))
    items.sort(key=lambda x: x[1])

    best_two_items = items[:2]
    best_two = [item[0] for item in best_two_items]
    best_two_clouds = [item[1] for item in best_two_items]

    return key, best_two, best_two_clouds


####################
#       主程序      #
####################

grouped_s2s3_data = match_pairs(filtered_sentinel23pairs_cloud_zip_files)

s23_filtered_grouped_data = {}
all_selected_clouds = []

task_args = [(key, value, oceans) for key, value in grouped_s2s3_data.items()]
total = len(task_args)

with concurrent.futures.ProcessPoolExecutor() as executor:
    for key, result, clouds in tqdm(executor.map(process_one_key, task_args),
                                    total=total,
                                    desc="Processing tiles"):
        if result is not None:
            s23_filtered_grouped_data[key] = result
            all_selected_clouds.extend(clouds)

# 计算平均云量
if all_selected_clouds:
    avg_cloud = sum(all_selected_clouds) / len(all_selected_clouds)
else:
    avg_cloud = None

print("有效 key 数量:", len(s23_filtered_grouped_data))
print("平均云量:", avg_cloud)

with open('s23_filtered_grouped_data.pkl', 'wb') as f:  # 注意是 'wb' 二进制写入
    pickle.dump(s23_filtered_grouped_data, f)
```

```python
import lmdb
import pickle

def read_lmdb(db_path, key=None):
    """
    读取LMDB数据库
    
    Args:
        db_path: LMDB数据库路径
        key: 要读取的键，如果为None则读取所有数据
    
    Returns:
        如果指定key，返回对应的值；如果不指定key，返回包含所有键值对的字典
    """
    env = lmdb.open(db_path, readonly=True, lock=False, readahead=False)
    
    with env.begin() as txn:
        if key is not None:
            # 读取单个键值对
            value = txn.get(key.encode() if isinstance(key, str) else key)
            if value is not None:
                try:
                    # 尝试反序列化（如果数据是用pickle序列化的）
                    return pickle.loads(value)
                except:
                    # 如果反序列化失败，返回原始字节数据
                    return value
            else:
                return None
        else:
            # 读取所有数据
            data = {}
            cursor = txn.cursor()
            for db_key, value in cursor:
                try:
                    data[db_key] = pickle.loads(value)
                except:
                    data[db_key] = value
            return data
```

```python
def get_all_keys(db_path):
    """
    获取LMDB数据库中的所有key
    
    Args:
        db_path: LMDB数据库路径
    
    Returns:
        key列表
    """
    env = lmdb.open(db_path, readonly=True, lock=False, readahead=False)
    
    keys = []
    with env.begin() as txn:
        cursor = txn.cursor()
        for key, _ in cursor:
            # 将bytes类型的key转换为字符串（如果key原本是字符串的话）
            try:
                keys.append(key.decode('utf-8'))
            except:
                keys.append(key)
    
    env.close()
    return keys


all_keys = get_all_keys(lmdb_path)
```
